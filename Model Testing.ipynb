```python
import numpy as np 
import pandas as pd 
import pickle
import json
```

```python
with open("linear_reg_med_ins.pkl", 'rb') as f: 
    model = pickle.load(f)
model
```

```python
with open("med_ins_column_data.json", "r") as f: 
    col_data = json.load(f)
col_data
```

```python
col_names = model.feature_names_in_ 
col_names
```

```python
age = 26
gender = "male"
bmi	= 31.435
children = 2 
smoker = "yes"
region = "southeast"

gender = col_data['gender'][gender] 
print("gender :",gender)

smoker = col_data['smoker'][smoker] 
print("smoker :",smoker)

region = f"region_{region}" 
print(region)
region_index = np.where(col_names == region)[0][0]
region_index
```

```python
test_array = np.zeros((1,model.n_features_in_)) 
test_array
```

```python
test_array[0,0] = age
test_array[0,1] = gender 
test_array[0,2] = bmi
test_array[0,3] = children 
test_array[0,4] = smoker
test_array[0,region_index] = 1 
test_array
```

```python
test_df = pd.DataFrame(test_array, columns = col_names) 
test_df
```

```python
pred_charges = np.around(model.predict(test_df)[0],3)
print("Predicted Medical Insurance Charges :",pred_charges)
```

```python
def test_med_ins_charges(age, gender, bmi, children, smoker, region): 
    gender = col_data['gender'][gender]
    smoker = col_data['smoker'][smoker] 
    region = f"region_{region}"
    region_index = np.where(col_names == region)[0][0]
    test_array = np.zeros((1,model.n_features_in_))
    
    test_array[0,0] = age
    test_array[0,1] = gender 
    test_array[0,2] = bmi
    test_array[0,3] = children 
    test_array[0,4] = smoker
    test_array[0,region_index] = 1
    
    test_df = pd.DataFrame(test_array, columns = col_names) 
    pred_charges = np.around(model.predict(test_df)[0],3) 
    return pred_charges

age = 35
gender = "female"
bmi	= 23.435
children = 1 
smoker = "no"
region = "southwest"

pred_charges = test_med_ins_charges(age, gender, bmi, children, smoker, region) 
print("Predicted Medical Insurance Charges :",pred_charges)


```

```python
age = 45
gender = "male"
bmi	= 45.435
children = 3 
smoker = "yes"
region = "northeast"

pred_charges = test_med_ins_charges(age, gender, bmi, children, smoker, region) 
print("Predicted Medical Insurance Charges :",pred_charges)
```
